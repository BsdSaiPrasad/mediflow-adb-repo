# Calendar Table

Builds the standard calendar dimension `adb_mediflow.silver.calendar` with one row per day from 1940-01-01 to 2030-12-31.

```sql
CREATE OR REPLACE TABLE adb_mediflow.silver.calendar AS
WITH dates AS (
  SELECT explode(sequence(DATE'1940-01-01', DATE'2030-12-31', INTERVAL 1 DAY)) AS date
)
SELECT
  CAST(date_format(date, 'yyyyMMdd') AS INT)          AS date_key,
  date,
  year(date)                                          AS year,
  quarter(date)                                       AS quarter,
  concat('Q', quarter(date))                          AS quarter_name,
  concat(year(date), '-Q', quarter(date))             AS year_quarter,
  month(date)                                         AS month,
  date_format(date, 'MMMM')                           AS month_name,
  date_format(date, 'MMM')                            AS month_short_name,
  date_format(date, 'yyyy-MM')                        AS year_month,
  weekofyear(date)                                    AS week_of_year,
  day(date)                                           AS day_of_month,
  dayofyear(date)                                     AS day_of_year,
  weekday(date) + 1                                   AS day_of_week,      -- 1 = Monday ... 7 = Sunday
  date_format(date, 'EEEE')                           AS day_name,
  date_format(date, 'EEE')                            AS day_short_name,
  weekday(date) >= 5                                  AS is_weekend,
  date_trunc('WEEK', date)::DATE                      AS week_start_date,
  date_add(date_trunc('WEEK', date)::DATE, 6)         AS week_end_date,
  trunc(date, 'MM')                                   AS month_start_date,
  last_day(date)                                      AS month_end_date,
  trunc(date, 'QUARTER')                              AS quarter_start_date,
  add_months(trunc(date, 'QUARTER'), 3) - 1           AS quarter_end_date,
  trunc(date, 'YEAR')                                 AS year_start_date,
  make_date(year(date), 12, 31)                       AS year_end_date,
  day(last_day(date))                                 AS days_in_month,
  (year(date) % 4 = 0 AND year(date) % 100 <> 0) OR year(date) % 400 = 0 AS is_leap_year
FROM dates
ORDER BY date;
```

```sql
SELECT count(*) AS row_count, min(date) AS min_date, max(date) AS max_date, count(DISTINCT date_key) AS distinct_keys
FROM adb_mediflow.silver.calendar;
```
